```python
df["Review Rating"]=df.groupby("Category")["Review Rating"].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(" ","_")
```

```python
df.columns
```

```python
df=df.rename(columns={"purchase_amount_(usd)":"purchase_amount"})
df=df.rename(columns={"promo_code_used":"Promo code_used"})
```

```python
df.columns
```

```python
#creating a age_group column
labels=["Young Adult","Adult","Middle_Age","Senior"]
df["age_group"]=pd.qcut(df["age"],q=4,labels=labels)
```

```python
df[["age","age_group"]].head()
```

```python
#creating purchase_frequency_days column
frequency_mapping={"Fortnightly":14,
                   "Weekly":7,
                   "Monthly":30,
                   "Quarterly":90,
                   "Bi-Weekly":14,
                   "Annually":365,
                   "Every 3 Months":90}
```

```python
df["purchase_frequency_days"]=df["frequency_of_purchases"].map(frequency_mapping)
```

```python
df[["purchase_frequency_days","frequency_of_purchases"]].head(15)
```

```python
df[["discount_applied","Promo code_used"]].head(15)
```

```python
(df["discount_applied"]==df["Promo code_used"]).all()
```

```python
df=df.drop("Promo code_used",axis=1)
```
